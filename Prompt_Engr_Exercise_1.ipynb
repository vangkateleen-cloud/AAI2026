```python
# Simulated Prompt Chain Execution

customer_message = "I was charged twice for my subscription this month."

# STEP 1 — Classification
def classify_issue(message):
    if "charg" in message.lower():
        return {"category": "billing", "reason": "duplicate charge mentioned"}
    return {"category": "other", "reason": "no billing keywords found"}

step1_output = classify_issue(customer_message)
print("STEP 1 OUTPUT (Classification):")
print(step1_output)

# STEP 2 — Gather Missing Info (dependent on category)
def gather_info(classification):
    category = classification["category"]
    if category == "billing":
        return [
            "What is the exact charge date?",
            "Are the transaction IDs identical?",
            "What are the last 4 digits of the card used?"
        ]
    return ["Please provide additional details about the issue."]

step2_output = gather_info(step1_output)
print("\nSTEP 2 OUTPUT (Follow-Up Questions - v1):")
for q in step2_output:
    print("-", q)

# ITERATION CYCLE — Improving Follow-Up Specificity
print("\n--- ITERATION: Improving Follow-Up Questions (v2) ---")

def improved_gather_info(classification):
    category = classification["category"]
    if category == "billing":
        return [
            "Confirm exact charge timestamps (include timezone).",
            "Provide screenshots of both transaction entries.",
            "Confirm whether charges appear pending or completed."
        ]
    return ["Provide detailed description including error messages."]

improved_questions = improved_gather_info(step1_output)
print("STEP 2 OUTPUT (Follow-Up Questions - v2 FINAL):")
for q in improved_questions:
    print("-", q)

# Simulated customer answers (based on v2 questions)
customer_answers = {
    "timestamps": "March 1, 2026 09:14 PT and March 1, 2026 09:16 PT",
    "screenshots": "Provided",
    "status": "Both completed"
}

# STEP 3 — Resolution (depends on step 1 + answers)
def propose_solution(classification, answers):
    if classification["category"] == "billing":
        return {
            "steps": [
                "Verify both transaction records in the billing system using timestamps and IDs.",
                "Confirm this is a duplicate charge (different transaction IDs within same billing cycle).",
                "Issue a refund for the duplicate charge and generate a confirmation/reference number.",
                "Notify the customer with the confirmation number and refund amount.",
                "Advise the refund may take 5–7 business days to appear on their statement."
            ],
            "escalation": "If refund fails or charges look suspicious, escalate to Payments Team with logs + screenshots."
        }
    return {
        "steps": ["Escalate to general support queue."],
        "escalation": "Immediate escalation required."
    }

step3_output = propose_solution(step1_output, customer_answers)

print("\nSTEP 3 OUTPUT (Resolution):")
for i, step in enumerate(step3_output["steps"], 1):
    print(f"{i}. {step}")

print("\nEscalation Rule:")
print(step3_output["escalation"])
```
