```python
# Stage A — REASON (in comments)
# We need k most frequent elements.
# Edge cases: k<=0, k>unique, empty list, non-list input.
# Tie-breaking: if two numbers have same frequency, return smaller value first.
#
# Stage B — PLAN (in comments)
# Count frequencies using collections.Counter.
# Sort items by (-frequency, value) to enforce deterministic ties.
# Return first k elements.
# Complexity: O(n + u log u), where u = number of unique elements.
#
# Stage C — ACT (Generate code) — Iteration 1: using Counter.most_common (has tie ambiguity)

from collections import Counter

def top_k_frequent_v1(nums, k):
    """Return k most frequent elements (v1). WARNING: tie-breaking may be nondeterministic."""
    if not isinstance(nums, list):
        raise TypeError("nums must be a list")
    if not isinstance(k, int):
        raise TypeError("k must be an int")
    if k <= 0:
        return []
    freq = Counter(nums)
    return [x for x, _ in freq.most_common(k)]  # tie-breaking not guaranteed

# Stage D — RUN tests for v1
print("=== Iteration 1 (v1) Tests ===")
print("Test A:", top_k_frequent_v1([1,1,2,2,2,3], 2), "Expected: [2, 1]")
print("Test B (tie case):", top_k_frequent_v1([3,3,2,2,1], 2), "Expected tie order: [2, 3]")
print("Test C (k>unique):", top_k_frequent_v1([1,2,2], 5), "Expected: [2, 1]")
print("Test D (k<=0):", top_k_frequent_v1([1,1,2], 0), "Expected: []")
```

```python
# Stage E — OBSERVE + FIX
# Observation: Counter.most_common may not guarantee deterministic ordering for ties.
# Fix: sort manually by (-frequency, value).

from collections import Counter
from typing import List

def top_k_frequent(nums: List[int], k: int) -> List[int]:
    """
    Return the k most frequent elements in nums.

    Rules:
    - If k <= 0: return []
    - If k > number of unique elements: return all unique elements
    - Ties broken deterministically by smaller integer first
    - Standard library only
    """
    if not isinstance(nums, list):
        raise TypeError("nums must be a list")
    if not isinstance(k, int):
        raise TypeError("k must be an int")
    if k <= 0:
        return []

    freq = Counter(nums)
    # Sort by frequency descending, then value ascending for deterministic ties
    items = sorted(freq.items(), key=lambda x: (-x[1], x[0]))
    return [val for val, _ in items[:k]]

# Re-run tests (execution evidence)
print("\n=== Iteration 2 (fixed) Tests ===")
print("Test A:", top_k_frequent([1,1,2,2,2,3], 2), "Expected: [2, 1]")
print("Test B (tie case):", top_k_frequent([3,3,2,2,1], 2), "Expected: [2, 3]")
print("Test C (k>unique):", top_k_frequent([1,2,2], 5), "Expected: [2, 1]")
print("Test D (k<=0):", top_k_frequent([1,1,2], 0), "Expected: []")

# Extra edge-case tests (robustness)
print("Test E (empty nums):", top_k_frequent([], 3), "Expected: []")
try:
    top_k_frequent("not a list", 2)
except Exception as e:
    print("Test F (invalid nums type):", type(e).__name__, str(e))

try:
    top_k_frequent([1,2,3], "2")
except Exception as e:
    print("Test G (invalid k type):", type(e).__name__, str(e))
```
